+ Sa se scrie o functie care primeste ca argument un vector si afiseaza doar elementele pare ale acestuia. 
+ Includeti comentarii si la inceput (creati doc-string) si intre instructiuni.

```python
# initializare lista vectori
vector = []

def vect_par(vector):
    ''' Se afiseaza elementele pare ale unui vector dat'''

    for i in range(len(vector)): # itereaza fiecare element
        if vector[i] % 2 == 0: # verificare a paritatii
            print(f"{vector[i]} este par")
```

Apelati functia pe un exemplu.
