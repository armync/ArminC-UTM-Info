+ Ex. 5

Scrieti o functie care primeste ca argument un ndarray si returneaza true daca acesta contine cel putin un numar intre 11-15 

```python
def verificareRange(vector):
    return np.any(vector>=11) & np.any(vector<=15)

numere = np.array([4, 23, 11, 13, 0])
```

Testati functia pentru cel putin doi vectori

```python
print(verificareRange(numere))
print(verificareRange(np.array([1,2])))
print(verificareRange(np.array([14])))
```

Sa se scrie un cod folosind functia where ce genereaza un ndarray cu elemente lui w modificate astfel: \
elementele intre valorile 20 si 50 sa fie trecute cu semnul -

```python
w = np.array([3, 0, 7, 21, 20, 17, 35, 44, 89])
```

```python
w_negativ = np.where((w>=20) & (w<=50), -w, w) # ultimul arg. reprezinta elementele ce nu indeplinesc verificarea
```

```python
print(w_negativ)
```
