# DataFrame

Un dataframe reprezintă un tabel de date cu linii si coloane, fiecare dintre acestea putând fi de tipuri diferite (numeric, șir, boolean etc.) 
 


DataFrame are un index pentru rânduri și pentru coloane; poate fi gândit ca un dict de Series, toate avand același index.


Putem construi un Dataframe plecand de la un dictionar de liste  

```python
data = {'Vant': ['T', 'F','T', 'T'], 'Temperatura': [15, 20, 7, 10], 'Umiditate': ['scazuta', 'normala', 'ridicata', 'normala']}
```

```python
data
```

```python
df = pd.DataFrame(data)
df
```

Putem preciza ordinea coloanelor. Putem adauga coloane noi. Cele fara valori apar cu NaN.

```python
df2 = pd.DataFrame(data, columns =['Temperatura', 'Vant', 'Umiditate', 'Clasa'])
df3 = pd.DataFrame(data, columns =['Temperatura', 'Vant', 'Umiditate'])
df2
```

```python
df2['Vant']
type(df2['Vant'])
```

```python
df2['Vant'][1] 
```

```python
df2.dtypes
```
