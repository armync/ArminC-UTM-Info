```python
def functie_2(v):
    """
    Functia calculeaza suma elementelor negative ale vectorului v sau 0, daca nu exista elemente negative in vector.
    
    """
    print("vectorul este", v)
    suma =0
    for i in v:
        if i<0:
            suma += i
    return suma

functie_2([1, -2, -3, 4])
        
    
```
