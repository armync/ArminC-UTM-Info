# Exemplu de imagine filtrata cu mai multe tipuri de kernel 



```python
from PIL import Image, ImageFilter  #PIL Python Image Library
```

```python
image = Image.open("UTM.jpg") 
```

```python
display(image)
```

```python
# kernel folosit pt edge detection
kernel = [
    -1, -1, -1,
    -1,  8, -1,
    -1, -1, -1
]
```

```python
#Obtinem imaginea filtrata cu filtrul de mai sus
filtered_image1 = image.filter(ImageFilter.Kernel((3, 3), kernel, scale = 1)) 
# scale = factorul la care se imparte fiecare pixel. 
# defUault: suma valorilor din filtru (kernel)
```

```python
display(filtered_image1)
```

```python
image = Image.open("ateneu.jpg") 
```

```python
# kernel folosit pt edge detection
kernel = [
    -1, -1, -1,
    -1,  8, -1,
    -1, -1, -1
]
```

```python
filtered_image2 = image.filter(ImageFilter.Kernel((3, 3), kernel, scale = 1)) 
 
```

```python
display(filtered_image2)
```

Incercam cu un alt filtru:

```python
kernel = [
    1, 2, 1,
    2, 4, 2,
    1, 2, 1
]
```

```python
filtered_image3 = image.filter(ImageFilter.Kernel((3, 3), kernel, scale = 3))
```

```python
display(filtered_image3)
```

```python
# kernel pentru sharpen image
kernel = [
    0, -1, 0,
    -1, 5, -1,
    0, -1, 0
]
```

```python
filtered_image4 = image.filter(ImageFilter.Kernel((3, 3), kernel, scale =1))
```

```python
display(filtered_image4)
```

```python
# kernel pentru box blur

kernel = [
    1, 1, 1,
    1, 1, 1,
    1, 1, 1
]
```

```python
filtered_image5 = image.filter(ImageFilter.Kernel((3, 3), kernel, scale = 1))
display(filtered_image5)
```

```python
filtered_image6 = image.filter(ImageFilter.GaussianBlur(radius=3))
```

```python
display(filtered_image6)
```
