```python
import re
import csv

def curata_tweeturi(tweeturi):
    tweeturi_curate = []
    
    for tweet in tweeturi:
        # hashtag-urile
        tweet = re.sub(r'#\w+', '', tweet)
        
        # @tweets
        tweet = re.sub(r'@\w+', '', tweet)
        
        # link
        tweet = re.sub(r'http\S+', '', tweet)
        
        # punctuatie si caractere speciale
        tweet = re.sub(r'[^\w\s]', '', tweet)
        
        # se adauga tweets curatat...
        tweeturi_curate.append(tweet)
    
    return tweeturi_curate

def scrie_in_csv(tweeturi_curate, nume_fisier):
    with open(nume_fisier, 'w', newline='', encoding='utf-8') as csvfile:
        fieldnames = ['Tweet Curat']
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        
        writer.writeheader()
        for tweet_curat in tweeturi_curate:
            writer.writerow({'Tweet Curat': tweet_curat})

# Exemplu de folosire
tweeturi_orig = [
    "tweet cu #hashtag interesant! @tweets",
    "Verificati acest link: https://twitter.com/.com",
    "Mesaj despre #eveniment? @utilizator",
    "Testam o trimitere >> 'wording'"
]
tweeturi_curate = curata_tweeturi(tweeturi_orig)

# Scrie datele curățate într-un fișier CSV
scrie_in_csv(tweeturi_curate, 'tweeturi_curate_1.csv')
print("rezultat in CSV ")
```

```python
import re
import csv
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# descarcare resurse (rulate o singura data pe masina)
for resursa in ['punkt', 'stopwords']:
    try:
        nltk.data.find(resursa)
    except LookupError:
        nltk.download(resursa, quiet=True)

# functie pentru curatarea tweeturilor
def curata_tweeturi(tweeturi, stop_lang='english'):
    sw = set(stopwords.words(stop_lang))
    rezultate = []
    for t in tweeturi:
        # scoate URL-urile
        t = re.sub(r'http\S+|www\.\S+', ' ', t)
        # tokenizeaza
        tokens = word_tokenize(t)
        # pastreaza doar alfanumerice, fara stopwords (case-insensitive) si fara #/@ la inceput
        tokens_curate = [
            tok for tok in tokens
            if tok.isalnum()
            and tok.lower() not in sw
            and not tok.startswith(('#', '@'))
        ]
        text_curat = ' '.join(tokens_curate)
        # normalizeaza spatiile
        text_curat = re.sub(r'\s+', ' ', text_curat).strip()
        rezultate.append(text_curat)
    return rezultate

# functie pentru scrierea rezultatelor intr-un CSV
def scrie_in_csv(linii, nume_fisier):
    with open(nume_fisier, 'w', newline='', encoding='utf-8') as f:
        w = csv.writer(f)
        w.writerow(['Tweet Curat'])
        for linie in linii:
            w.writerow([linie])

# exemplu de utilizare
if __name__ == "__main__":
    tweeturi_orig = [
        "tweet cu #hashtag interesant! @tweets",
        "Verificati acest link: https://twitter.com/.com",
        "Mesaj despre #eveniment? @utilizator",
        "Testam o trimitere >> 'wording'"
    ]

    rez = curata_tweeturi(tweeturi_orig, stop_lang='english')  # poti pune 'romanian' daca vrei
    scrie_in_csv(rez, 'tweeturi_curate_nltk_bun_2.csv')
    print("OK. Fisierul 'tweeturi_curate_nltk_bun_2.csv' a fost generat cu succes!")
```

```python
# importuri necesare
import re
import csv
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import TweetTokenizer

# adauga o cale suplimentara pentru resursele NLTK (optional)
nltk.data.path.append('')

# descarca pachetele necesare (fara mesaje suplimentare)
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

# tokenizer pentru texte tip tweet
_tknzr = TweetTokenizer(strip_handles=True, reduce_len=True, preserve_case=False)

# set combinat de stopwords engleza + romana
_stop_words = set(stopwords.words('english')) | set(stopwords.words('romanian'))

# expresie regulata pentru eliminarea URL-urilor
_url_re = re.compile(r'https?://\S+|www\.\S+', re.IGNORECASE)

def curata_tweeturi(tweeturi):
    """
    Curata o lista de tweet-uri.
    - Elimina URL-uri, mentiuni si caractere speciale
    - Elimina stopwords si pastreaza doar cuvinte alfanumerice
    - Normalizeaza textul la lowercase
    """
    tweeturi_curate = []

    for tweet in tweeturi:
        # eliminare link-uri
        t = _url_re.sub('', tweet)

        # tokenizare
        tokens = _tknzr.tokenize(t)

        cuvinte_curate = []
        for tok in tokens:
            if tok.startswith('#'):
                tok = tok[1:]  # eliminare simbol hashtag
            if not tok or tok in _stop_words:
                continue
            if tok.isalnum():  # pastreaza doar cuvinte alfanumerice
                cuvinte_curate.append(tok)

        # reconstruieste textul curatat
        tweet_curat = ' '.join(cuvinte_curate).strip()
        tweeturi_curate.append(tweet_curat)

    return tweeturi_curate


def scrie_in_csv(tweeturi_curate, nume_fisier):
    """
    Scrie lista de tweet-uri curatate intr-un fisier CSV cu o singura coloana.
    """
    with open(nume_fisier, 'w', newline='', encoding='utf-8') as csvfile:
        fieldnames = ['Tweet Curat']
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        for tweet_curat in tweeturi_curate:
            writer.writerow({'Tweet Curat': tweet_curat})


# exemplu de utilizare
tweeturi_orig = [
    "tweet cu #hashtag interesant! @tweets",
    "Verificati acest link: https://twitter.com/.com",
    "Mesaj despre #eveniment? @utilizator",
    "Testam o trimitere >> 'wording'"
]

tweeturi_curate = curata_tweeturi(tweeturi_orig)
scrie_in_csv(tweeturi_curate, 'tweeturi_curate_nltk_4.csv')
print("Rezultatul a fost scris in fisierul CSV.")
```

```python
# -*- coding: utf-8 -*-
# Program demo: curățare text, tokenizare + frecvențe (NLTK),
# analiză TextBlob și traducere în română (deep-translator)

import os
import re
import nltk
from pathlib import Path
from nltk.tokenize import RegexpTokenizer
from nltk.probability import FreqDist
from textblob import TextBlob
from deep_translator import GoogleTranslator

# resursele NLTK/TextBlob
def ensure_nltk_data():
    preferred = Path("/databricks/driver/nltk_data")
    fallback = Path.home() / "nltk_data"
    data_dir = preferred if preferred.parent.exists() else fallback
    data_dir.mkdir(parents=True, exist_ok=True)

    # adaugă calea în NLTK
    if str(data_dir) not in nltk.data.path:
        nltk.data.path.append(str(data_dir))

    # resurse necesare
    resources = {
        "tokenizers/punkt": "punkt",  # pentru sentence/word tokenization (TextBlob)
        "taggers/averaged_perceptron_tagger": "averaged_perceptron_tagger",  # POS tagging
        "corpora/wordnet": "wordnet",          # pentru anumite funcții TextBlob
        "corpora/brown": "brown",              # folosit de POS tagger/np extractor
        "corpora/omw-1.4": "omw-1.4",          # sinonime/lematizare extinsă
        "corpora/stopwords": "stopwords",      # stopwords
    }

    for key, pkg in resources.items():
        try:
            nltk.data.find(key)
        except LookupError:
            nltk.download(pkg, download_dir=str(data_dir), quiet=True)

ensure_nltk_data()

# ---- date de intrare ----
text_original = "#AI @user http://example.com este super! Vezi educația? 😊"
text_cu_greseli = "I havv goood speling in engleza."

print("=== APLICAȚIE DE ANALIZĂ TEXT ===")
print(f"Text original: {text_original}\n")

# ---- PAS 1: curățare text ----
def curata_text(tweet: str) -> str:
    """Elimină hashtag-uri, mențiuni, link-uri, punctuație; normalizează spațiile; lowercase."""
    tweet = re.sub(r'#\w+', ' ', tweet)       # #hashtag
    tweet = re.sub(r'@\w+', ' ', tweet)       # @user
    tweet = re.sub(r'http\S+|www\.\S+', ' ', tweet)  # link-uri
    tweet = re.sub(r'[^\w\s]', ' ', tweet)    # punctuație/emoticoane
    tweet = re.sub(r'\s+', ' ', tweet).strip()
    return tweet.lower()

text_curat = curata_text(text_original)
print("1) Text curățat:")
print("   ", text_curat, "\n")

# ---- PAS 2: tokenizare + frecvențe cu NLTK (fără punkt) ----
tokenizer = RegexpTokenizer(r"\w+")  # păstrează doar cuvinte (fără punctuație)
tokens = tokenizer.tokenize(text_curat)
print("2) Tokeni extrași:", tokens)

freq_dist = FreqDist(tokens)
print("   Top frecvențe:")
for w, f in freq_dist.most_common(5):
    print(f"   - {w}: {f}")
print()

# ---- PAS 3: analiză cu TextBlob (are nevoie de punkt & tagger) ----
blob = TextBlob(text_original)
print("3) Analiză cu TextBlob (pe textul original):")
print("   - Words:", list(blob.words))
print("   - Sentences:", [str(s) for s in blob.sentences])
print("   - Noun phrases:", list(blob.noun_phrases))
print("   - POS tags:", blob.tags)
print("   - Sentiment (polarity, subjectivity):", blob.sentiment)

# Traducere în română (folosim deep-translator în loc de blob.translate)
try:
    traducere_ro = GoogleTranslator(source="auto", target="ro").translate(str(blob))
    print("   - Traducere în română:", traducere_ro)
except Exception as e:
    print("   - Traducere în română: [a eșuat]", e)

# ---- PAS 4: corectare ortografică cu TextBlob ----
blob_greseli = TextBlob(text_cu_greseli)
try:
    corectat = blob_greseli.correct()  # funcționează pe texte engleze
except Exception as e:
    corectat = f"[corectarea a eșuat: {e}]"

print("\n4) Corectare ortografică:")
print("   Original:", text_cu_greseli)
print("   Corectat:", corectat)

print("\n=== Sfârșit aplicație ===")
```
