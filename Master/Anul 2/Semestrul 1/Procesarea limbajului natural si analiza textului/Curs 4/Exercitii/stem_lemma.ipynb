## Stemming Vs Lemmatization

```python
# Importam bibliotecile necesare pentru procesarea limbajului natural (NLP)
import nltk

# Tokenizarea inseamna impartirea unui text in unitati mai mici numite "tokeni" (de obicei cuvinte sau semne de punctuatie)
nltk.download('punkt')  # Descarcam pachetul 'punkt' care contine modele pre-antrenate pentru tokenizare
from nltk.tokenize import word_tokenize  # Importam functia pentru a imparti textul in cuvinte


# POS (Part-of-Speech) tagging inseamna etichetarea fiecarui cuvant cu rolul sau gramatical (ex: substantiv, verb, adjectiv etc.)
from nltk import pos_tag  # Importam functia pentru etichetarea partilor de vorbire


# WordNet foloseste un set diferit de etichete fata de POS tagging-ul standard din NLTK.
# Aceasta etapa va fi utila pentru a converti etichetele POS in formatul folosit de WordNet pentru lematizare.
from nltk.corpus import wordnet  # Importam corpusul WordNet pentru lucrul cu sinonime, antonime si lematizare

# Stopwords sunt cuvinte comune (precum "si", "in", "este") care de obicei sunt eliminate deoarece nu aduc valoare semantica.
nltk.download('stopwords')  # Descarcam lista de cuvinte "stop"
from nltk.corpus import stopwords  # Importam setul de cuvinte comune pentru filtrare ulterioara

# Stemming este procesul de reducere a unui cuvant la radacina sa (ex: "running" -> "run")
# Este o metoda bazata pe reguli simple, care nu garanteaza un cuvant valid din punct de vedere gramatical.
from nltk.stem import PorterStemmer  # Importam algoritmul Porter Stemmer, unul dintre cei mai folositi stemmeri

# Lemmatizarea este un proces mai avansat decat stemming-ul.
# Ea transforma cuvintele la forma lor de baza (lemma) tinand cont de context si de rolul gramatical.
# De exemplu: "better" -> "good", "running" -> "run".
from nltk.stem import WordNetLemmatizer  # Importam clasa pentru lematizare folosind baza de date WordNet
```

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import stopwordsiso as stopwordsiso
import snowballstemmer



text_en = """Each synset represents a distinct semantic concept and contains logical relations between words, 
such as synonyms and antonyms. In the context of working with the NLTK library, WordNet is mainly used 
for lemmatization, which means transforming words to their base dictionary form (lemma). 
For example, the word running becomes run, and better becomes good. This is possible because 
WordNet understands the grammatical category and lexical meaning of the word, simplifying semantic analysis."""

print("=== TEXT IN ENGLEZA ===")
print(text_en)
print()

# 1. Tokenizare (impartirea textului in cuvinte)
tokens_en = word_tokenize(text_en)
print("=== TOKENIZARE (EN) ===")
print(tokens_en)
print()

# 2. Eliminare stopwords (cuvinte comune)
stop_words_en = set(stopwords.words('english'))
filtered_en = [word for word in tokens_en if word.lower() not in stop_words_en and word.isalpha()]
print("=== DUPA ELIMINAREA STOPWORDS (EN) ===")
print(filtered_en)
print()

# 3. Stemming (reducerea cuvintelor la radacina)
stemmer_en = PorterStemmer()
stemmed_en = [stemmer_en.stem(word) for word in filtered_en]
print("=== STEMMING (EN) ===")
print(stemmed_en)
print()



text_ro = """Fiecare synset reprezinta un concept semantic distinct si contine relatii logice intre cuvinte, 
cum ar fi sinonime, antonime etc. In contextul lucrului cu biblioteca NLTK, WordNet este folosit in special 
pentru lematizare, adica pentru transformarea cuvintelor la forma lor de baza corecta din dictionar (lemma). 
De exemplu, cuvantul running devine run, iar better devine good. Acest lucru este posibil deoarece WordNet 
intelege categoria gramaticala si sensul lexical al cuvantului, ceea ce simplifica analiza semantica a textului."""

print("TEXT IN ROMANA ")
print(text_ro)
print()

# 1. Tokenizare (impartirea textului in cuvinte)
tokens_ro = word_tokenize(text_ro)
print("TOKENIZARE RO")
print(tokens_ro)
print()

# 2. Eliminare stopwords (folosim biblioteca stopwordsiso pentru limba romana)
stop_words_ro = stopwordsiso.stopwords("ro")
filtered_ro = [word for word in tokens_ro if word.lower() not in stop_words_ro and word.isalpha()]
print("DUPA ELIMINAREA STOPWORDS RO")
print(filtered_ro)
print()

# 3. Stemming (folosim SnowballStemmer pentru limba romana)
stemmer_ro = snowballstemmer.stemmer('romanian')
stemmed_ro = stemmer_ro.stemWords(filtered_ro)
print("=== STEMMING (RO) ===")
print(stemmed_ro)
print()
```

```python
# Imports
import nltk
nltk.download('averaged_perceptron_tagger')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk import pos_tag
from nltk.stem import WordNetLemmatizer, PorterStemmer

#  mapare POS NLTK -> WordNet
def to_wn_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    return wordnet.NOUN

text = """Each synset represents a distinct semantic concept and contains logical relations between words, such as synonyms and antonyms. In the context of working with the NLTK library, WordNet is mainly used 
for lemmatization, which means transforming words to their base dictionary form (lemma). 
For example, the word running becomes run, and better becomes good. This is possible because 
WordNet understands the grammatical category and lexical meaning of the word, simplifying semantic analysis."""

print("TEXT ORIGINAL")
print(text, "\n")

#  Tokenizare pe tot textul
tokens = word_tokenize(text)

#  POS tagging pe toti tokenii (pentru context mai bun)
pos_tags = pos_tag(tokens)

#  Afisare parti de vorbire (doar tokeni alfabetici)
print(" ETICHETARE POS (toti tokenii alfabetici)")
for tok, tag in pos_tags:
    if tok.isalpha():
        print(f"{tok:<15} -> {tag}")
print()

# Lemmatizare cu POS corect (pe lowercase)
lemmatizer = WordNetLemmatizer()
lemmas = []
for tok, tag in pos_tags:
    if tok.isalpha():  # ignoram punctuatia
        wn_pos = to_wn_pos(tag)
        lemmas.append(lemmatizer.lemmatize(tok.lower(), wn_pos))

print("")
print(lemmas, "\n")

# Stopwords (pastram negatiile importante)
stop_words = set(stopwords.words('english'))
for keep in ("not", "no", "nor"):
    if keep in stop_words:
        stop_words.remove(keep)

lemmas_nostop = [w for w in lemmas if w not in stop_words]
print("DUPA ELIMINAREA STOPWORDS (pe leme)")
print(lemmas_nostop, "\n")

# Stemming (pentru comparatie) – aplicat peste lemele fara stopwords
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in lemmas_nostop]
print("STEMMING (dupa leme si fara stopwords) ")
print(stems)

```

```python
from nltk import pos_tag, word_tokenize

text = """Each synset represents a distinct semantic concept and contains logical relations between words, such as synonyms and antonyms. In the context of working with the NLTK library, WordNet is mainly used 
for lemmatization, which means transforming words to their base dictionary form (lemma). 
For example, the word running becomes run, and better becomes good. This is possible because 
WordNet understands the grammatical category and lexical meaning of the word, simplifying semantic analysis."""

tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)

# Filtram doar tokenii alfabetici
alpha_tokens = [tok for tok, tag in pos_tags if tok.isalpha()]

# Afisam lungimea maxima
max_len = max(len(tok) for tok in alpha_tokens)
print(f"Cuvantul cel mai lung are {max_len} caractere.\n")

# Afisam aliniat in functie de lungimea maxima
for tok, tag in pos_tags:
    if tok.isalpha():
        print(f"{tok:<{max_len}} -> {tag}")
```

```python
import spacy
from spacy.cli import download as spacy_download
import re
import pandas as pd
from nltk.stem.snowball import SnowballStemmer
from wordcloud import WordCloud
import matplotlib.pyplot as plt

FILE_PATH = "input.txt"
LANG = "ro"  # ex: ro, en, fr, de, es, it, pt, nl
OUTPUT_CSV = "nlp_output.csv"
print(FILE_PATH, LANG, OUTPUT_CSV)

LANG_TO_MODEL = {
    "ro": "ro_core_news_sm",
    "en": "en_core_web_sm",
    "fr": "fr_core_news_sm",
    "de": "de_core_news_sm",
    "es": "es_core_news_sm",
    "it": "it_core_news_sm",
    "pt": "pt_core_news_sm",
    "nl": "nl_core_news_sm",
}

def ensure_spacy_model(lang_code: str) -> str:
    # Returnează modelul potrivit limbii alese
    return LANG_TO_MODEL.get(lang_code, "xx_sent_ud_sm")

def load_nlp(lang_code: str):
    name = ensure_spacy_model(lang_code)
    try:
        return spacy.load(name)
    except OSError:
        print(f"Instalez modelul {name}...")
        spacy_download(name)
        return spacy.load(name)

nlp = load_nlp(LANG)
print("Model:", nlp.meta.get("name"), "lang:", LANG)

# Configurare stemmer pentru limbă
STEM_LANG_MAP = {
    "ro": "romanian", "en": "english", "fr": "french", "de": "german",
    "es": "spanish", "it": "italian", "pt": "portuguese", "nl": "dutch"
}
def get_stemmer(lang_code: str):
    # Unstemmer potrivit pentru limbă
    code = STEM_LANG_MAP.get(lang_code, "english")
    try:
        return SnowballStemmer(code)
    except Exception:
        return None

stemmer = get_stemmer(LANG)

# Funcție simplă de curățare a textului
def basic_clean(t: str) -> str:
    # Transformă textul în lowercase și elimină caractere inutile
    t = t.lower()
    t = re.sub(r"[\r\t]", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t

# Citirea textului din fișierul .txt
with open(FILE_PATH, "r", encoding="utf-8", errors="ignore") as f:
    raw_text = f.read()

clean_text = basic_clean(raw_text)
print(clean_text[:300])

# Procesare NLP: tokenizare, lematizare, POS, stem, frecvență
doc = nlp(clean_text)
records = []
spacy_stop = getattr(nlp.Defaults, "stop_words", set())

def valid_token(tok) -> bool:
    # Excludere spațiile, semne de punctuație și stopwords
    if tok.is_space or tok.is_punct or tok.like_num:
        return False
    if tok.text in spacy_stop:
        return False
    return True

# Extragere informații utile pentru fiecare token valid
for tok in doc:
    if not valid_token(tok):
        continue
    w = tok.text
    le = tok.lemma_ or w
    pos = tok.pos_ or "X"
    if stemmer:
        try:
            st = stemmer.stem(w)
        except Exception:
            st = w
    else:
        st = w
    records.append((w, le, st, pos))

# Creare DataFrame și calcul frecvențe
df = pd.DataFrame(records, columns=["cuvant", "lema", "stema", "parte_de_vorbire"])
if df.empty:
    print("Nu există tokeni validați.")
else:
    freq = (
        df.value_counts(["cuvant", "lema", "stema", "parte_de_vorbire"])
          .reset_index(name="frecventa")
          .sort_values("frecventa", ascending=False)
          .reset_index(drop=True)
    )
    try:
        display(freq.head(20))
    except NameError:
        print(freq.head(20))

# Salvare rezultate în CSV
if not df.empty:
    freq.to_csv(OUTPUT_CSV, index=False, encoding="utf-8")
    print("CSV:", OUTPUT_CSV)
else:
    print("Nu s-a salvat CSV (fără date).")

# Afișare frecvențe pe părți de vorbire (POS)
if not df.empty:
    pos_freq = (
        freq.groupby("parte_de_vorbire")["frecventa"]
            .sum().reset_index()
            .sort_values("frecventa", ascending=False)
    )
    try:
        display(pos_freq)
    except NameError:
        print(pos_freq)
else:
    print("Fără date POS.")

# Creare word cloud doar pentru verbe
if not df.empty:
    verbs = freq[freq["parte_de_vorbire"] == "VERB"]
    if not verbs.empty:
        lemma_counts = verbs.groupby("lema")["frecventa"].sum().to_dict()
        wc = WordCloud(width=1000, height=600, background_color="white").generate_from_frequencies(lemma_counts)
        plt.figure(figsize=(10, 6))
        plt.imshow(wc, interpolation="bilinear")
        plt.axis("off")
        plt.title("Word Cloud – verbe (leme)")
        plt.show()
    else:
        print("Nu s-au găsit verbe.")
else:
    print("Fără date pentru word cloud.")
```
