```python
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

text = "Exista mai multe pachete Python utile pentru extragerea a mai multe cuvinte cheie din texte"
tokens = word_tokenize(text)
freq_dist = FreqDist(tokens)
keywords = freq_dist.most_common(5)

descriptions = {
    "exista": "Acesta este un verb care arata prezenta.",
    "mai": "Marca de grad comparativ.",
    "multe": "Numeral care indica o cantitate.",
    "pachete": "Un set de fisiere Python.",
    "Python": "Un limbaj de programare versatil si popular.",
    "utile": "Indica utilitatea.",
}

for keyword, frequency in keywords:
    description = descriptions.get(keyword, "Fara descriere disponibila.")
    print(f"Cuvant cheie: {keyword}, Frecventa: {frequency}, Descriere: {description}")
```

```python
sorted_keywords = sorted(keywords, key=lambda x: x[1], reverse=True)

for keyword, frequency in sorted_keywords:
    description = descriptions.get(keyword, "lipsa descriere.")
    print(f"Cuvant cheie: {keyword}, Frecventa: {frequency}, Descriere: {description}")
```
