## RegexpTokenizer(r'\w+') împarte textul doar în secvențe alfanumerice (ignoră semnele de punctuație)

```python
from nltk.tokenize import RegexpTokenizer
from nltk.probability import FreqDist

text = "Machine learning și inteligența artificială sunt domenii care transformă lumea modernă. Asta este un fapt!!!!;; Se va schimba oare ceva,,?"
tokenizer = RegexpTokenizer(r'\w+')
tokens = tokenizer.tokenize(text)
freq_dist = FreqDist(tokens)

for word, freq in freq_dist.most_common(5):
    print(word, freq)
```

```python
import nltk
from nltk.tokenize import sent_tokenize, RegexpTokenizer
from nltk.probability import FreqDist

text = "Python-ul. este un limbaj de programare care s-a dezvoltat încă din anii 2000! Un lucru foarte interesant. Este foarte popular în analiza datelor! Îți place și ție Python?"

# Împărțim textul în propoziții
sentences = sent_tokenize(text)

# Tokenizer care împarte doar pe cuvinte (fără semne de punctuație)
tokenizer = RegexpTokenizer(r'\w+')

# Afișăm propozițiile, cuvintele și frecvența lor
for i, sentence in enumerate(sentences, 1):
    print(f"\nPropoziția {i}: {sentence}")
    
    words = tokenizer.tokenize(sentence)
    freq_dist = FreqDist(words)   # distribuția de frecvență a cuvintelor
    
    for j, word in enumerate(words, 1):
        print(f"   {j}. {word} (frecvență: {freq_dist[word]})")
```

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.probability import FreqDist

# doar prima dată 
nltk.download('punkt', quiet=True)

text = "Python este un limbaj de programare. Este foarte popular în analiza datelor! Îți place și ție Python?"

#  propoziții
sentences = sent_tokenize(text)

# propoziții, cuvintele, frecvența
for i, sentence in enumerate(sentences, 1):
    print(f"\nPropoziția {i}: {sentence}")
    
    words = word_tokenize(sentence)
    freq_dist = FreqDist(words)   # frecvență a cuvintelor
    
    for j, word in enumerate(words, 1):
        print(f"   {j}. {word} (frecvență: {freq_dist[word]})")
```

```python
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Text de analizat
text = "Inteligența artificială transformă multiple industrii prin automatizare predicție, iar modele avansate sunt folosite în sănătate educație finanțe îmbunătățind eficiența și acuratețea deciziilor."

# Tokenizarea textului și calculul frecvenței cuvintelor
tokens = word_tokenize(text)
freq_dist = FreqDist(tokens)
keywords = freq_dist.most_common(5)  # primele 5 cuvinte cheie

# Dicționar de descrieri asociate cuvintelor cheie
descriptions = {
    "Inteligența artificială": "Tehnologie care imită capacitățile cognitive umane.",
    "Inteligența": "Capacitatea de a învăța și de a rezolva probleme.",
    "artificială": "Creată de om, care nu apare natural.",
    "transformă": "Schimbă sau îmbunătățește într-un mod semnificativ.",
    "multiple": "Mai multe, numeroase.",
    "industrii": "Sectore economice sau domenii de activitate.",
    "automatizare": "Procesul de reducere a intervenției umane prin tehnologie.",
    "predicție": "Estimarea unui rezultat viitor pe baza datelor actuale.",
    "eficiență": "Realizarea unui rezultat optim cu resurse minime."
}

# Sortarea cuvintelor cheie după frecvență în ordine descrescătoare
sorted_keywords = sorted(keywords, key=lambda x: x[1], reverse=True)

# Afișarea cuvintelor cheie și descrierilor în ordine descrescătoare
for keyword, frequency in sorted_keywords:
    description = descriptions.get(keyword, "lipsa descriere.")
    print(f"Cuvânt cheie: {keyword}, Frecvență: {frequency}, Descriere: {description}")
```

```python
from textblob import TextBlob

def extract_keywords(text):
    # Obiect TextBlob
    blob = TextBlob(text)
    
    # Extrage expresiile nominale
    keywords = blob.noun_phrases
    
    return keywords

def main():
    #text = "Inteligența artificială joacă un rol tot mai important în domenii precum sănătate, educație și transporturi. Algoritmii de machine learning și rețelele neuronale permit analiza rapidă a datelor și oferă soluții precise pentru probleme complexe."

    text = ("The Artificial Intelligence plays a greater role in fields such as healthcare, education and transportation. Machine learning and neural networks algorithms allow rapid analysis of the data and offer precise solutions to complex issues.")
    
    # Extrage cuvintele cheie
    keywords = extract_keywords(text)
    
    print("Textul de analizat:")
    print(text)
    print("\nCuvintele cheie identificate:::")
    for keyword in keywords:
        print("- " + keyword)

if __name__ == "__main__":
    main()
```
