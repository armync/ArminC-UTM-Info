```python
import csv
import googleapiclient.discovery
import googleapiclient.errors

api_service_name = "youtube"
api_version = "v3"
DEVELOPER_KEY = "AIzaSyAkggDIMf4TZsNXPHioOuJU2SIllAwAJlU" # del this

youtube = googleapiclient.discovery.build(
    api_service_name, api_version, developerKey=DEVELOPER_KEY)

video_id = "e7ho6z32yyo"  
csv_filename = 'video_comments_.csv'


with open(csv_filename, 'w', newline='', encoding='utf-8') as csv_file:
    
    fieldnames = ['Comment', 'Author', 'Published At', 'Updated At', 'Like Count']

  
    writer = csv.DictWriter(csv_file, fieldnames=fieldnames)

    # Scrie antetul în fișierul CSV
    writer.writeheader()

    next_page_token = None

    while True:
        
        request = youtube.commentThreads().list(
            part="snippet",
            videoId=video_id,
            maxResults=100,
            pageToken=next_page_token
        )
        response = request.execute()

        for item in response.get('items', []):
            comment = item['snippet']['topLevelComment']['snippet']
            comment_data = {
                'Comment': comment['textDisplay'],
                'Author': comment['authorDisplayName'],
                'Published At': comment['publishedAt'],
                'Updated At': comment['updatedAt'],
                'Like Count': comment['likeCount']
            }
            writer.writerow(comment_data)

        next_page_token = response.get('nextPageToken')

        if not next_page_token:
            break

print("Comentariile au fost scrise în", csv_filename)
```

```python
import spacy
import ipywidgets as widgets # pentru interfață interactivă
from IPython.display import display, HTML

nlp = spacy.load("ro_core_news_sm")

def reply(message: str) -> str:
    text = message.upper()
    if "NLP" in text or "PROCESARE LIMBAJ NATURAL" in text:
        return ("Procesarea Limbajului Natural (NLP) este o ramură a inteligenței artificiale "
                "care se ocupă de interacțiunea dintre calculatoare și limbajul uman.")
    elif "ENTITATE" in text:
        return ("În procesarea limbajului natural, o entitate poate fi orice element "
                "cu semnificație, cum ar fi o persoană, un loc sau un obiect.")
    elif "ANALIZĂ SENTIMENT" in text or "ANALIZA SENTIMENT" in text:
        return ("Analiza sentimentului este o tehnică în NLP care determină emoția asociată "
                "unui text, cum ar fi pozitivă, negativă sau neutră.")
    elif "TOKENIZARE" in text:
        return ("Tokenizarea constă în descompunerea unui text în unități mai mici, numite token-uri, "
                "cum ar fi cuvinte sau simboluri.")
    elif "LEMATIZARE" in text:
        return ("Lematizarea implică reducerea unui cuvânt la forma sa de bază, de exemplu, "
                "transformarea 'alergam' în 'alerga'.")
    elif "STEMMING" in text or "RĂDĂCINARE" in text:
        return ("Stemmingul este procesul de reducere a unui cuvânt la rădăcina sa, "
                "prin tăierea sufixelor, fără a garanta o formă corectă din punct de vedere lingvistic.")
    elif "MODELE PRE-ANTRENATE" in text or "MODELE PREANTRENATE" in text:
        return ("Modelele pre-antrenate în spaCy sau Hugging Face sunt modele de limbă antrenate pe volume mari de text "
                "și pot fi folosite pentru diverse sarcini NLP, precum recunoașterea entităților sau analiza sentimentului.")
    elif "WORD EMBEDDINGS" in text or "VECTORIZARE" in text:
        return ("Word embeddings sunt reprezentări numerice ale cuvintelor într-un spațiu vectorial, "
                "unde cuvintele cu semnificații similare sunt apropiate între ele.")
    elif "TRANSFORMER" in text or "BERT" in text or "GPT" in text:
        return ("Arhitecturile de tip Transformer, precum BERT sau GPT, folosesc mecanismul de atenție "
                "pentru a înțelege contextul cuvintelor într-un text și sunt fundamentale în NLP modern.")
    elif "RECOAȘTERE VOCALĂ" in text or "SPEECH TO TEXT" in text:
        return ("Recunoașterea vocală transformă vorbirea în text folosind modele acustice și lingvistice, "
                "fiind o componentă importantă în aplicațiile de asistenți vocali.")
    elif "CHATBOT" in text or "ASISTENT VIRTUAL" in text:
        return ("Chatboturile sunt aplicații NLP care pot conversa cu utilizatorii, "
                "folosind reguli predefinite sau modele de limbaj antrenate.")
    elif "TRADUCERE AUTOMATĂ" in text or "MACHINE TRANSLATION" in text:
        return ("Traducerea automată folosește modele NLP pentru a converti textul dintr-o limbă în alta, "
                "cu ajutorul rețelelor neuronale și al modelelor Transformer.")
    elif "SPACY" in text:
        return ("spaCy este o bibliotecă NLP puternică pentru Python, care oferă instrumente pentru tokenizare, "
                "analiză gramaticală, recunoașterea entităților și multe altele.")
    else:
        return "Mulțumesc pentru întrebare. Cum te pot ajuta?"


title = HTML("<h3>Chatbot</h3>")
input_box = widgets.Text(
    placeholder="Scrie mesajul tău...",
    layout=widgets.Layout(width="80%")
)
send_btn = widgets.Button(description="Trimite")
reset_btn = widgets.Button(description="Resetează conversația")
output = widgets.HTML()
log = []  # Istoricul conversației

# Funcție pentru a afișa conversația în format HTML
def render():
    html = []
    for sender, msg in log:
        if sender == "Tu":
            html.append(f"<p><b>{sender}:</b> {msg}</p>")
        else:
            html.append(f"<div style='background:#f2f2f2;padding:8px;border-radius:6px;'><b>{sender}:</b> {msg}</div>")
    output.value = "".join(html)

def on_send(_):
    text = input_box.value.strip()
    if not text:
        return
    log.append(("Tu", text))  # Adaugă mesajul utilizatorului în istoric
    log.append(("Chatbot", reply(text)))  # Adaugă răspunsul chatbotului
    input_box.value = ""  # Golește câmpul de input
    render()  # Actualizează afișarea conversației

def on_reset(_):
    log.clear() # Reset
    render()

send_btn.on_click(on_send)
reset_btn.on_click(on_reset)

render()
display(title, input_box, send_btn, reset_btn, output)
```
