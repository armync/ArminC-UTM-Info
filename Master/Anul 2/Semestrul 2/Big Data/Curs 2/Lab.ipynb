```sql
%python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, LongType, TimestampType
import random
import time

# Initialize Spark session
spark = SparkSession.builder.appName("UsersDirtyDuplicates").getOrCreate()

# Schema
schema = StructType([
    StructField("user_id", StringType(), True),
    StructField("user_first_touch_timestamp", LongType(), True),
    StructField("email", StringType(), True),
    StructField("updated", TimestampType(), True)
])

# Sample emails (valid + invalid + missing)
emails = [
    "alice@example.com", "bob@example.com", "carol@example.com",
    "dave@example.com", "eve@sample.org", "", None,
    "invalid_email", "frank@example.com", "grace@sample.org"
]

num_rows = 30000
sample_data = []

for i in range(1, num_rows + 1):
    # 7% chance to create duplicate user_id
    if random.random() < 0.07:
        user_id = f"user_{random.randint(1, num_rows//2)}"
    else:
        user_id = f"user_{i}"
    
    # Random timestamp in last ~2 years (microseconds)
    timestamp_us = int((time.time() - random.randint(0, 2*365*24*3600)) * 1_000_000)
    
    # 5% chance to duplicate (user_id, timestamp) tuple
    if random.random() < 0.05 and len(sample_data) > 0:
        prev_row = random.choice(sample_data)
        user_id = prev_row[0]
        timestamp_us = prev_row[1]
    
    # Random email
    if random.random() < 0.05 and len(sample_data) > 0:
        email = random.choice([row[2] for row in sample_data if row[2]])
    else:
        email = random.choice(emails)
    
    # Append row with updated=None initially
    sample_data.append((user_id, timestamp_us, email, None))

# Create DataFrame
users_dirty_df = spark.createDataFrame(sample_data, schema)

# Fill updated column with current timestamp
users_dirty_df = users_dirty_df.withColumn(
    "updated",
    F.current_timestamp()
)

# Save as table in Databricks
#spark.sql("USE SCHEMA `utm-master-an2-sd&ia-dl`")
#spark.sql("USE CATALOG `utm-master-an2-sd&ia-dl.seturi_de_date`")

#users_dirty_df.write.mode("overwrite").saveAsTable("`utm-master-an2-sd&ia-dl`.`seturi_de_date`.users_dirty")

# Create database if not exists
#spark.sql("CREATE DATABASE IF NOT EXISTS `seturi_de_date`")

#we comute schema (database)
#spark.sql("USE `seturi_de_date`")
#users_dirty_df.write.mode("overwrite").saveAsTable("`seturi_de_date`.users_dirty")


# Save the table in catalog/schema
users_dirty_df.write.format("delta").mode("overwrite").saveAsTable("`utm-master-an2-sd&ia-dl`.`claudia`.`users_dirty`")

### nu merge sa salvez in seturi_de_date
# Salvează tabelul în catalog/schema
#users_dirty_df.write.format("delta").write.mode("overwrite").saveAsTable("`utm-master-an2-sd&ia-dl`.`seturi_de_date`.users_dirty")




#users_dirty_df.write.format("delta").mode("overwrite").saveAsTable#("users_dirty")

#the table users_dirty is saved as a managed table in your current default schema (database) and catalog. By default, this is usually the default schema in the workspace's default catalog, unless you have previously changed it with a USE statement. The data is stored in the Databricks workspace's managed storage, and the table is registered in the metastore

#You can check your current schema and catalog with:
#display(spark.sql("SELECT current_catalog(), current_database()"))

# Show sample
users_dirty_df.show(10, truncate=False)



```


## Inspect Missing Data

Based on the counts above, it looks like there are at least a handful of null values in some of our fields.

**NOTE:** Null values behave incorrectly in some math functions, including **`count()`**.

- **`count(col)`** skips **`NULL`** values when counting specific columns or expressions.
- **`count(*)`** is a special case that counts the total number of rows (including rows that are only **`NULL`** values).

We can count null values in a field by filtering for records where that field is null, using either:  
**`count_if(col IS NULL)`** or **`count(*)`** with a filter for where **`col IS NULL`**. 

Both statements below correctly count records with missing emails.

```sql
SELECT count_if(email IS NULL) FROM users_dirty;
SELECT count(*) FROM users_dirty WHERE email IS NULL;
```

```sql
SELECT count(*) FROM users_dirty WHERE email IS NULL;
```

```sql
SELECT count_if(email IS NULL) FROM users_dirty;
```

The code below uses **`GROUP BY`** to remove duplicate records based on **`user_id`** and **`user_first_touch_timestamp`** column values. (Recall that these fields are both generated when a given user is first encountered, thus forming unique tuples.)

Here, we are using the aggregate function **`max`** as a hack to:
- Keep values from the **`email`** and **`updated`** columns in the result of our group by
- Capture non-null emails when multiple records are present

```sql
CREATE OR REPLACE TEMP VIEW deduped_users AS 
SELECT user_id, user_first_touch_timestamp, max(email) AS email, max(updated) AS updated
FROM users_dirty
WHERE user_id IS NOT NULL
GROUP BY user_id, user_first_touch_timestamp;

SELECT count(*) FROM deduped_users
```

```sql
%python
from pyspark.sql.functions import max
dedupedDF = (usersDF
    .where(col("user_id").isNotNull())
    .groupBy("user_id", "user_first_touch_timestamp")
    .agg(max("email").alias("email"), 
         max("updated").alias("updated"))
    )

dedupedDF.count()
```


 
## Date Format and Regex
Now that we've removed null fields and eliminated duplicates, we may wish to extract further value out of the data.

The code below:
- Correctly scales and casts the **`user_first_touch_timestamp`** to a valid timestamp
- Extracts the calendar date and clock time for this timestamp in human readable format
- Uses **`regexp_extract`** to extract the domains from the email column using regex

```sql
SELECT *, 
  date_format(first_touch, "MMM d, yyyy") AS first_touch_date,
  date_format(first_touch, "HH:mm:ss") AS first_touch_time,
  regexp_extract(email, "(?<=@).+", 0) AS email_domain
FROM (
  SELECT *,
    CAST(user_first_touch_timestamp / 1e6 AS timestamp) AS first_touch 
  FROM deduped_users
)
```

This SQL query performs data wrangling and data enrichment on the deduped_users table. It transforms raw fields into more usable, analytics-friendly columns.

1️⃣ Inner query – timestamp conversion
SELECT *,
  CAST(user_first_touch_timestamp / 1e6 AS timestamp) AS first_touch
FROM deduped_users

What happens here:

user_first_touch_timestamp is a numeric timestamp, typically stored in microseconds

Dividing by 1e6 converts it to seconds

CAST(... AS timestamp) converts it into a readable timestamp

➡ This creates a new column called first_touch, representing the user’s first interaction date and time.

2️⃣ Outer query – data enrichment
SELECT *,
  date_format(first_touch, "MMM d, yyyy") AS first_touch_date,
  date_format(first_touch, "HH:mm:ss") AS first_touch_time,
  regexp_extract(email, "(?<=@).+", 0) AS email_domain
FROM (...)

a) date_format(first_touch, "MMM d, yyyy")

Extracts only the date

Formats it in a human-readable way (e.g. Jan 5, 2023)

b) date_format(first_touch, "HH:mm:ss")

Extracts only the time

Format: hours:minutes:seconds

c) regexp_extract(email, "(?<=@).+", 0)

Extracts the email domain

Example:

john.doe@gmail.com → gmail.com

3️⃣ Final result

For each user, the output includes:

All original columns

A readable timestamp of the first interaction

The date and time split into separate columns

The email domain extracted for analysis

4️⃣ Why this is useful

✔ Makes raw data easier to analyze
✔ Enables grouping by date, time, or email domain
✔ Prepares data for analytics, reporting, or ML models

5️⃣ Type of operation

This query is an example of:

Data wrangling

Data transformation

Data enrichment

Summary

The query converts raw timestamps into readable dates and enriches the dataset with derived fields that make downstream analysis simpler and more effective.

```sql
%python
from pyspark.sql.functions import date_format, regexp_extract

display(dedupedDF
    .withColumn("first_touch", (col("user_first_touch_timestamp") / 1e6).cast("timestamp"))
    .withColumn("first_touch_date", date_format("first_touch", "MMM d, yyyy"))
    .withColumn("first_touch_time", date_format("first_touch", "HH:mm:ss"))
    .withColumn("email_domain", regexp_extract("email", "(?<=@).+", 0))
)
```
