```python
%sql
-- Selectează catalogul 'samples'
USE CATALOG samples;
```

```python
%sql
-- Afișează lista tabelelor disponibile din schema 'nyctaxi'
SHOW TABLES IN nyctaxi;
```

```python
%sql
-- Alege tabelul 'nyctaxi.trips' și afișează primele 10 înregistrări
SELECT * FROM nyctaxi.trips LIMIT 10;
```

```python
# Încarcă tabelul ales într-un Spark DataFrame
df = spark.table("samples.nyctaxi.trips")

# Afișează schema DataFrame-ului
df.printSchema()

# Afișează primele 5 rânduri din DataFrame
display(df.limit(5))
```

```python
# Realizează o filtrare a datelor (curse cu distanța mai mare de 5 mile) și creează un nou DataFrame
df_filtrat = spark.sql("""
SELECT * FROM samples.nyctaxi.trips
WHERE trip_distance > 5
""")

# Afișează rezultatul obținut
display(df_filtrat)
```

```python
# Salvează DataFrame-ul rezultat într-o tabelă Delta din schema proprie
df_filtrat.write.format("delta").mode("overwrite").saveAsTable(
    "`utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_over5mi`"
)
```

```python
%sql
-- Verifică existența tabelului creat folosind comenzi SQL
SHOW TABLES IN `utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`;
```

```python
%sql
-- Afișează conținutul tabelului creat
SELECT * FROM `utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_over5mi` LIMIT 10;
```

```python
%sql
-- Determină numărul total de înregistrări
SELECT COUNT(*) AS total_inregistrari FROM `utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_over5mi`;
```

```python
%sql

-- Creează o tabelă derivată folosind CREATE TABLE AS SELECT
CREATE OR REPLACE TABLE `utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_cash` AS
SELECT *
FROM `utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_over5mi`
WHERE fare_amount > 20;
```

```python
%sql
-- Identifică sursele upstream pentru tabelul creat
DESCRIBE HISTORY `utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_cash`;
```

```python
%sql
-- Tabelul sursă 'samples.nyctaxi.trips' a fost interogat și filtrat în notebook, rezultatul fiind salvat ca tabel Delta '`utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_over5mi`'.
-- Ulterior, s-a creat o tabelă derivată '`utm-master-an2-sd&ia-dl`.`armin_andrei_chanchian`.`nyctaxi_trips_cash`' pe baza filtrării suplimentare a datelor.
```
