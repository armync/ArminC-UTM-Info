```python
data = [
    (1, "Student 1", "IT", 5000),
    (2, " Student 2", "HR", 4200),
    (3, " Student 3", "IT", 6000),
    (4, " Student 4", "Finance", 5500)
]
columns = ["id", "nume", "departament", "salariu"]
df = spark.createDataFrame(data, columns)
df.show()
```

```python
df.write.format("delta").saveAsTable(
    "`utm-master-an2-sd&ia-dl`.mironela.angajati"
)

```

```python
%sql
SELECT * 
FROM `utm-master-an2-sd&ia-dl`.mironela.angajati;
```

Comanda CREATE TABLE AS SELECT creează datele,dar nu le afișează. Pentru a vedea datele, trebuie un SELECT separat

```python
%sql
CREATE TABLE `utm-master-an2-sd&ia-dl`.mironela.angajati_it AS
SELECT *
FROM `utm-master-an2-sd&ia-dl`.mironela.angajati
WHERE departament = 'IT';
```

```python
%sql
SHOW CATALOGS;
USE CATALOG `utm-master-an2-sd&ia-dl`;

SHOW SCHEMAS;
USE SCHEMA mironela;

SHOW TABLES;

SELECT * FROM angajati;
SELECT * FROM angajati_it;
```

```python
%sql
CREATE TABLE studenti_sql (
    id INT,
    nume STRING,
    departament STRING,
    salariu INT
)
USING DELTA;
```

```python
%sql
INSERT INTO studenti_sql VALUES
(1, 'Student 1', 'IT', 5000),
(2, 'Student 2', 'HR', 4200),
(3, 'Student 3', 'IT', 6000);
```

Un tabel Delta este un tabel Big Data stocat în Data Lake, care folosește tehnologia Delta Lake pentru a adăuga funcționalități avansate peste fișierele clasice (Parquet).
Delta Table = fișiere în Data Lake + reguli care le fac să se comporte ca un tabel „inteligent”.
Delta Lake este un strat software care se află: peste Data Lake (S3, ADLS, DBFS), sub Spark / Databricks

```python
%sql DESCRIBE EXTENDED studenti_sql;
```
