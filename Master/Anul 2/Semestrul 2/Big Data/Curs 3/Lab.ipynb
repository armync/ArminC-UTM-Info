```python
%sql
describe samples.tpcds_sf1000.store_sales;
```

```python
%sql
SELECT 
  ss_store_sk,
  SUM(ss_quantity) AS total_quantity
FROM samples.tpcds_sf1000.store_sales
GROUP BY ss_store_sk;
```

```python
%sql
SELECT 
  SUM(ss_quantity) AS total_quantity
FROM samples.tpcds_sf1000.store_sales
WHERE ss_quantity > 10;
```

```python
%sql
SELECT COUNT(ss_quantity)
FROM samples.tpcds_sf1000.store_sales;
```

```python
%sql
SELECT 
  ss_store_sk,
  ss_item_sk,
  SUM(ss_quantity) AS total_quantity
FROM samples.tpcds_sf1000.store_sales
GROUP BY ss_store_sk, ss_item_sk;
```

```python
%sql
SELECT 
  ss_store_sk,
  SUM(ss_quantity) AS total_quantity
FROM samples.tpcds_sf1000.store_sales
GROUP BY ss_store_sk
ORDER BY total_quantity DESC
LIMIT 10;
```
